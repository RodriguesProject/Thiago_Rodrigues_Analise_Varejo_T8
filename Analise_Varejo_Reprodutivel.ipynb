```python
from pathlib import Path
from IPython.display import display, Image
from analise_varejo import carregar_base, preparar_dados, gerar_relatorio_qualidade, gerar_tabelas, analisar_dataframe

CAMINHO_BASE = Path('Base Varejo.csv')
if not CAMINHO_BASE.exists():
    try:
        from google.colab import files
        enviados = files.upload()
        CAMINHO_BASE = Path(next(iter(enviados)))
    except ImportError as exc:
        raise FileNotFoundError('Ajuste CAMINHO_BASE para o CSV local.') from exc

base = carregar_base(CAMINHO_BASE)
print(base.shape)
display(base.head())
```

## Qualidade e preparação

Os campos originais são preservados. Duplicidades são reportadas, não removidas automaticamente.

```python
qualidade = gerar_relatorio_qualidade(base)
display(qualidade)
preparado = preparar_dados(base)
tabelas = gerar_tabelas(preparado)
for nome, tabela in tabelas.items():
    print(nome)
    display(tabela)
```

## Exportação e visualizações

Os arquivos são gerados localmente em `outputs/`. As contagens sobre linhas representam registros de itens, não clientes ou compras.

```python
resumo = analisar_dataframe(base, 'outputs')
print(resumo)
for nome in ['registros_por_categoria.png', 'registros_por_segmento.png']:
    caminho = Path('outputs') / nome
    if caminho.exists():
        display(Image(filename=str(caminho)))
```

## Limitações

A análise é descritiva. Não calcula faturamento, margem ou ticket médio porque as colunas documentadas não incluem preços/custos. A execução de testes com amostras sintéticas não substitui a execução da base original completa.
