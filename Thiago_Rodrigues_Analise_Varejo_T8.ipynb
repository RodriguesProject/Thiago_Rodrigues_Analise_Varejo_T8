## 3. Etapa 3 - Limpeza

```python
# Converter de int64 para date
df['DATA'] = pd.to_datetime(
    df['DATA'],
    format='%d/%m/%Y'
).dt.date
```

```python
df.info()
```

```python
# Mapeamento dos códigos do Estado Civil para seus significados descritivos
mapeamento_ec = {
    1: 'Casado ou união estável',
    2: 'Divorciado',
    3: 'Separado',
    4: 'Solteiro',
    5: 'Viúvo'
}

# Substitui os valores numéricos pelas strings correspondentes
df['CL_EC'] = df['CL_EC'].map(mapeamento_ec).astype(str)

# Exibe a distribuição dos novos valores e as primeiras linhas do DataFrame
print("Distribuição do Estado Civil:")
print(df['CL_EC'].value_counts())
print("\nTipos de dados atualizados:")
print(df[['CL_EC']].info())

display(df.head())
```

```python
# Substitui a denominação '#N/D' na coluna PR_CAT por 'ND'
df['PR_CAT'] = df['PR_CAT'].replace('#N/D', 'ND')

# Exibe a contagem de valores na coluna para validar a alteração
print("Valores atualizados na coluna PR_CAT:")
print(df['PR_CAT'].value_counts(dropna=False))
```

## 4. Etapa 4 - Funções


```python
def obter_top3_coluna(dataframe, coluna):
    """
    Retorna os top 3 valores mais frequentes de uma coluna,
    suas contagens e a representatividade percentual.
    """
    contagem = dataframe[coluna].value_counts()
    percentual = dataframe[coluna].value_counts(normalize=True) * 100

    top3 = pd.DataFrame({
        'Valor': contagem.index[:3],
        'Frequência': contagem.values[:3],
        'Percentual (%)': percentual.values[:3].round(2)
    })
    return top3

# Gerar o top 3 para cada uma das colunas do DataFrame
for col in df.columns:
    print("="*50)
    print(f"TOP 3 DA COLUNA: {col}")
    print("="*50)
    display(obter_top3_coluna(df, col))
    print("\n")
```

```python
def obter_bottom3_coluna(dataframe, coluna):
    """
    Retorna os 3 valores menos frequentes (inverso do top 3) de uma coluna,
    suas contagens e a representatividade percentual.
    """
    # Ordenamos de forma crescente (do menos frequente para o mais frequente)
    contagem = dataframe[coluna].value_counts(ascending=True)
    percentual = dataframe[coluna].value_counts(ascending=True, normalize=True) * 100

    bottom3 = pd.DataFrame({
        'Valor': contagem.index[:3],
        'Frequência': contagem.values[:3],
        'Percentual (%)': percentual.values[:3].round(4)
    })
    return bottom3

# Gerar o bottom 3 (valores menos comuns) para cada uma das colunas do DataFrame
for col in df.columns:
    print("="*50)
    print(f"TOP 3 MENOS COMUM DA COLUNA: {col}")
    print("="*50)
    display(obter_bottom3_coluna(df, col))
    print("\n")
```

## 5. Análise


```python
# =====================================================================
# ETAPA 5 — FILTROS E AGRUPAMENTOS (ESTRITAMENTE CÓDIGO)
# =====================================================================
import pandas as pd

# Aplicando o filtro focado nos clientes do segmento socioeconômico 'A'
df_classe_a = df[df['CL_SEG'] == 'A']

# Resumo 1: Contagem absoluta de registros (itens individuais vendidos) por categoria
resumo_categoria = df_classe_a.groupby('PR_CAT').size().reset_index(name='Total_Itens_Vendidos')
resumo_categoria = resumo_categoria.sort_values(by='Total_Itens_Vendidos', ascending=False)

# Resumo 2: Compras únicas (Notas Fiscais) e Clientes únicos por Gênero
resumo_genero = df_classe_a.groupby('CL_GENERO').agg(
    Clientes_Unicos=('CL_ID', 'nunique'),
    Compras_Unicas=('CO_ID', 'nunique'),
    Itens_Totais=('CO_ID', 'count')
).reset_index()

# Exibição dos resultados estruturados na tela
print("=== RESUMO 1: Contagem de Registros por Categoria (Segmento A) ===")
display(resumo_categoria)

print("\n=== RESUMO 2: Compras Únicas (Notas) e Clientes Únicos por Gênero (Segmento A) ===")
display(resumo_genero)
```

# **Relatório de Análise Geral do Dataset e Conclusão**

Este relatório apresenta o diagnóstico completo do comportamento de compras a partir do processamento e limpeza efetuados na base de dados de varejo.

---

## **1. Dimensão e Estrutura dos Dados**

* **Volume Geral:** O banco de dados registra **830.000 linhas** (itens adicionados ao carrinho).
* **Transações vs. Clientes:** Foram contabilizadas **18.471 compras únicas** (notas fiscais/cupons emitidos), divididas entre **1.000 clientes únicos** cadastrados.
* **Comportamento de Cesta:** O mesmo cliente retorna recorrentemente para realizar compras no estabelecimento (média de ~18,5 compras por cliente). Além disso, cada cupom fiscal de compra carrega, em média, **44,9 itens**, indicando um perfil forte de "compras de abastecimento/mês".

---

## **2. Perfil Comportamental e Demográfico do Consumidor**

* **Gênero:** Há um equilíbrio quase perfeito na base de clientes. No recorte efetuado para a **Classe A** (Segmento A), por exemplo, identificou-se **40 clientes mulheres** (gerando 700 notas fiscais) e **44 clientes homens** (gerando 792 notas fiscais).
* **Estado Civil:** Os clientes estão distribuídos de forma muito equilibrada entre as categorias civis. Solteiros, Casados, Divorciados e Separados mantêm um patamar similar de cerca de ~190k a ~210k registros cada, enquanto os viúvos representam a menor fatia com ~23k registros.
* **Núcleo Familiar (Filhos):** O número médio de filhos por cliente é de **1,15**, variando entre 0 e 4 filhos. A mediana é 0, o que nos diz que a maioria dos clientes registrados não possui filhos.

---

## **3. Mix de Produtos e Preferências**

* **Domínio de Alimentos:** A categoria **Alimentos** é a líder absoluta com **434.767 registros** (mais de 50% de todo o volume operado na base).
* **Outras Categorias:** Higiene (~155k) e Limpeza (~145k) aparecem como as principais categorias secundárias. Bebidas, Pet e Acessórios completam a cauda longa de vendas.
* **Saneamento Efetuado:** Identificou-se e removeu-se a inconsistência de texto `#N/D` na coluna de categorias (`PR_CAT`), normalizando 3.650 registros para a sigla limpa `ND`.

---

## **Conclusão Estratégica**

A base de dados de varejo analisada possui excelente consistência e qualidade (zero nulos estruturais). O perfil de consumo é amplamente dominado por produtos essenciais (Alimentos), servindo a um público de classe média/popular que realiza compras volumosas de reposição periódica.

A alta fidelidade dos clientes (1.000 clientes gerando 18 mil cupons de compra com média de ~45 itens por transação) indica que o estabelecimento é o canal principal de abastecimento dessas famílias.

**Recomendação:** Campanhas promocionais e posicionamento de gôndolas devem focar em estratégias de *cross-selling* (venda cruzada), estimulando que o cliente de Alimentos preencha seu carrinho volumoso também com os produtos de Higiene e Limpeza, que são as outras categorias de maior relevância na operação.
